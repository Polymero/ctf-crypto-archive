```python
import gzip
from Crypto.Cipher import AES
```

```python
gzip.compress(b'123456' * 6)
```

```python
KEY = b"HELLO AM KEY :3!"

def SecureSerialise(data):
    data = gzip.compress(data)
    aes = AES.new(KEY, AES.MODE_CTR, nonce=b"\x00"*12)
    return aes.encrypt(data).hex()

def SecureUnserialise(data):
    aes = AES.new(KEY, AES.MODE_CTR, nonce=b"\x00"*12)
    data = aes.decrypt(bytes.fromhex(data))
    return gzip.decompress(data)
```

```python
ser = SecureSerialise(b'123456' * 6)
SecureUnserialise(ser)
```

```python
SecureUnserialise("2ab97b2b84e7123182a92f06602440e516589921e053f2b92143db444e")
```

```python
for i in range(len(ser)*4):
    x = int(ser, 16) ^ 2**i
    y = "{:0{n}x}".format(x, n=-(-x.bit_length()//4))
    try:
        print(SecureUnserialise(y))
    except:
        print('error')
```
